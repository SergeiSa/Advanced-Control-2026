```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters
x = np.linspace(-1.4, 1.4, 2000)
x_cert = 1/np.sqrt(2)
x_roa = 1

# System and Lyapunov functions
f = -x + x**3
V = 0.5 * x**2
Vdot = x * f
minusV = -V

fontSize = 16
lineWidth = 2.5

# ================================================================
# FIGURE 1 - V(x) and system dynamics
# ================================================================

fig1, ax1 = plt.subplots(figsize=(10, 6))
fig1.patch.set_facecolor('white')

ax1.plot(x, V, linewidth=lineWidth, color='blue', label='V(x)')
ax1.set_xlabel('x', fontsize=fontSize)
ax1.set_ylabel('V(x)', fontsize=fontSize, color='blue')
ax1.tick_params(axis='y', labelcolor='blue')

ax2 = ax1.twinx()
ax2.plot(x, f, '--', linewidth=lineWidth, color='red', label='f(x)')
ax2.set_ylabel('f(x) = x_dot', fontsize=fontSize, color='red')
ax2.tick_params(axis='y', labelcolor='red')

ax1.axvline(0, color='black', linewidth=1.2)
ax1.axvline(-x_roa, color='black', linestyle=':', linewidth=1.5, label='True ROA')
ax1.axvline(x_roa, color='black', linestyle=':', linewidth=1.5)
ax1.axvline(-x_cert, color='black', linestyle='--', linewidth=1.5, label='Certified region')
ax1.axvline(x_cert, color='black', linestyle='--', linewidth=1.5)

ax1.set_title('V(x) and system dynamics x_dot = f(x)', fontsize=fontSize)
ax1.legend(loc='best', fontsize=fontSize-1)
ax1.grid(True)
ax1.set_xlim([-1.4, 1.4])
plt.tight_layout()
plt.show()

# ================================================================
# FIGURE 2 - Vdot(x) and system dynamics
# ================================================================

fig3, ax3 = plt.subplots(figsize=(10, 6))
fig3.patch.set_facecolor('white')

ax3.plot(x, Vdot, linewidth=lineWidth, color='blue', label='Vdot(x)')
ax3.set_xlabel('x', fontsize=fontSize)
ax3.set_ylabel('Vdot(x)', fontsize=fontSize, color='blue')
ax3.tick_params(axis='y', labelcolor='blue')

ax4 = ax3.twinx()
ax4.plot(x, f, '--', linewidth=lineWidth, color='red', label='f(x)')
ax4.set_ylabel('f(x) = x_dot', fontsize=fontSize, color='red')
ax4.tick_params(axis='y', labelcolor='red')

ax3.axhline(0, color='black', linewidth=1.0)
ax3.axvline(-x_roa, color='black', linestyle=':', linewidth=1.5, label='True ROA')
ax3.axvline(x_roa, color='black', linestyle=':', linewidth=1.5)
ax3.axvline(-x_cert, color='black', linestyle='--', linewidth=1.5, label='Certified region')
ax3.axvline(x_cert, color='black', linestyle='--', linewidth=1.5)

ax3.set_title('Vdot(x) and system dynamics x_dot = f(x)', fontsize=fontSize)
ax3.legend(loc='best', fontsize=fontSize-1)
ax3.grid(True)
ax3.set_xlim([-1.4, 1.4])
plt.tight_layout()
plt.show()

# ================================================================
# FIGURE 3 - V(x), -V(x), and Vdot(x) showing Lyapunov inequality
# ================================================================

fig5, ax5 = plt.subplots(figsize=(10, 6))
fig5.patch.set_facecolor('white')

ax5.plot(x, V, linewidth=lineWidth, label='V(x)', color='blue')
ax5.plot(x, minusV, '--', linewidth=lineWidth, label='-V(x)', color='green')
ax5.plot(x, Vdot, '-.', linewidth=lineWidth, label='Vdot(x)', color='red')

ax5.axhline(0, color='black', linewidth=1.0)
ax5.axvline(-x_cert, color='black', linestyle='--', linewidth=1.5, label='Certified region')
ax5.axvline(x_cert, color='black', linestyle='--', linewidth=1.5)
ax5.axvline(-x_roa, color='black', linestyle=':', linewidth=1.5, label='True ROA')
ax5.axvline(x_roa, color='black', linestyle=':', linewidth=1.5)

ax5.set_xlabel('x', fontsize=fontSize)
ax5.set_ylabel('V, -V, Vdot', fontsize=fontSize)
ax5.set_title('Lyapunov inequality: Vdot <= -V', fontsize=fontSize)

ax5.legend(loc='best', fontsize=fontSize-1)
ax5.grid(True)
ax5.set_xlim([-1.4, 1.4])
plt.tight_layout()
plt.show()

# ================================================================
# Display numerical values
# ================================================================

print(f'Certified region: |x| < {x_cert:.6f}')
print(f'True ROA:         |x| < {x_roa:.6f}')
print(f'V at certificate boundary: {0.5*x_cert**2:.6f}')
print(f'V at true ROA boundary:    {0.5*x_roa**2:.6f}')
```

```python
import numpy as np
from scipy.integrate import solve_ivp
import matplotlib.pyplot as plt

# Define the ODE: dx/dt = -x
def ode(t, x):
    return -x

# Solve from t=0 to t=5, with initial condition x(0)=1
sol = solve_ivp(ode, (0, 5), [1.0], t_eval=np.linspace(0, 5, 100))

# Plot the solution
plt.plot(sol.t, sol.y[0])
plt.xlabel('t')
plt.ylabel('x(t)')
plt.title(r'Solution of $\dot{x} = -x$')
plt.grid(True)
plt.show()
```

Find a region of attraction using direct Lyapunov method: propose a Lyapunov function candidate and show that:
1) it is positive on a region $S$: $V(x) > 0$ for all $x \in S$, $x \neq 0$, and

2) its derivative is negative on a region $S$: $V(x) < 0$ for all $x \in S$, $x \neq 0$

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.linalg import solve_continuous_are
from scipy.integrate import odeint

# Define the nonlinear function
def f(x):
    return 5 * np.sin(x) + x**4

# Linearization around x0 = 0
x0 = 0
delta = 0.001  # Small perturbation for numerical derivative
a = (f(x0 + delta) - f(x0)) / delta
print(f"Linearized system: a = {a:.4f}")

# LQR design for the linearized system
# System: \dot{x} = a*x + u
# Cost: J = ∫(q*x^2 + r*u^2) dt

q = 1.0  # State weight
r = 1.0  # Control weight

# Solve continuous-time algebraic Riccati equation
A = np.array([[a]])
B = np.array([[1.0]])
Q = np.array([[q]])
R = np.array([[r]])

# Solve Riccati equation: A'P + PA - PBR^{-1}B'P + Q = 0
P = solve_continuous_are(A, B, Q, R)
print(f"Riccati solution P = {P[0,0]:.4f}")

# Compute LQR gain: K = R^{-1} * B' * P
K = np.linalg.inv(R) @ B.T @ P
k = K[0, 0]  # Scalar gain
print(f"LQR gain K = {k:.4f}")

# Closed-loop nonlinear system with control u = -k*x
def closed_loop_nonlinear(x, t):
    """Closed-loop nonlinear system: dot{x} = f(x) - k*x"""
    return f(x) - k * x

# Define the Lyapunov function from Riccati solution
def lyapunov_function(x):
    """V(x) = x' * P * x"""
    return P[0, 0] * x**2

# Derivative of Lyapunov function along trajectories
def lyapunov_derivative(x):
    """dV/dt = dV/dx * dx/dt = 2*P*x * (f(x) - k*x)"""
    dxdt = f(x) - k * x
    return 2 * P[0, 0] * x * dxdt

# Generate x values for plotting
x_values = np.linspace(-2, 2.5, 1000)

# Plot Lyapunov function and its derivative
plt.figure(figsize=(10, 5))

# Subplot 1: Lyapunov function V(x)
plt.subplot(1, 2, 1)
V_values = [lyapunov_function(x) for x in x_values]
plt.plot(x_values, V_values, 'b-', linewidth=2, label='V(x) = x$^T$Px')
plt.xlabel('x')
plt.ylabel('V(x)')
plt.title('Lyapunov Function')
plt.grid(True, alpha=0.3)
plt.legend()
plt.axhline(y=0, color='k', linestyle='-', alpha=0.2)

# Subplot 2: Derivative of Lyapunov function
plt.subplot(1, 2, 2)
dV_values = [lyapunov_derivative(x) for x in x_values]
plt.plot(x_values, dV_values, 'r-', linewidth=2, label='dV/dt')
plt.xlabel('x')
plt.ylabel('dV/dt')
plt.title('Derivative of Lyapunov Function')
plt.grid(True, alpha=0.3)
plt.legend()
plt.axhline(y=0, color='k', linestyle='-', alpha=0.2)

plt.tight_layout()
plt.show()

# Also plot both on the same graph
plt.figure(figsize=(10, 5))
plt.plot(x_values, V_values, 'b-', linewidth=2, label='V(x)')
plt.plot(x_values, np.array(V_values) * -1, 'b:', linewidth=1, label='-V(x)')
plt.plot(x_values, dV_values, 'r-', linewidth=2, label='dV/dt')
plt.xlabel('x')
plt.ylabel('Value')
plt.title('Lyapunov Function and Its Derivative')
plt.grid(True, alpha=0.3)
plt.legend()
plt.axhline(y=0, color='k', linestyle='-', alpha=0.2)
plt.tight_layout()
plt.show()

# Simulate the closed-loop system to verify stability
def simulate_system(x0, t_span):
    """Simulate closed-loop system from initial condition"""
    t = np.linspace(0, t_span, 1000)
    sol = odeint(closed_loop_nonlinear, x0, t)
    return t, sol

# Simulate from different initial conditions
initial_conditions = [-2.0, -1.0, 1.0, 2.0]

plt.figure(figsize=(10, 5))

for i, x0_init in enumerate(initial_conditions):
    t, sol = simulate_system(x0_init, 5)

    plt.subplot(2, 2, i+1)
    plt.plot(t, sol, 'b-', linewidth=2)
    plt.xlabel('Time (s)')
    plt.ylabel('x(t)')
    plt.title(f'Trajectory from x(0) = {x0_init:.1f}')
    plt.grid(True, alpha=0.3)
    plt.axhline(y=0, color='k', linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()

# Compute and display numerical values
print("\n" + "="*50)
print("SUMMARY")
print("="*50)
print(f"System: f(x) = 5sin(x) + x")
print(f"Linearized about x0 = {x0}")
print(f"Linearization coefficient a = {a:.4f}")
print(f"LQR parameters: q = {q}, r = {r}")
print(f"Riccati solution P = {P[0,0]:.4f}")
print(f"LQR gain K = {k:.4f}")
print(f"Closed-loop system: dot{{x}} = 5sin(x) + x - {k:.4f}x")
print("\nNote: Since dV/dt is negative definite (except at x=0),")
print("the origin is asymptotically stable in the sense of Lyapunov.")
```
