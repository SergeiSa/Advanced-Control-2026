# Problem 1. Local Stability and Linear Feedback Stabilization of Polynomial Systems

For each system below, complete the following tasks.

---

## Task 1 — Local stability of the origin (open loop, $u = 0$)

Consider the system with $u = 0$:

$$
\left\{
\begin{aligned}
\dot{x}_1 &= f_1(x_1, x_2) \\
\dot{x}_2 &= f_2(x_1, x_2)
\end{aligned}
\right.
$$

1. Compute the Jacobian matrix
   $$
   A = \left.\frac{\partial f}{\partial x}\right|_{x=0}
   $$
2. Compute the eigenvalues of $A$.
3. Using the eigenvalue criterion, determine whether the origin $x_1 = 0,\ x_2 = 0$ is:
   - asymptotically stable,
   - marginally stable,
   - or unstable.

---

## Task 2 — Linear feedback stabilization

Introduce a linear control law of the form

$$
u = -k_1 x_1 - k_2 x_2
$$

1. Substitute $u$ into the system to obtain the closed-loop dynamics.
2. Compute the closed-loop Jacobian $A_c$ at the origin.
3. Choose gains $k_1, k_2$ (if possible) such that $A_c$ is Hurwitz, i.e. all eigenvalues have strictly negative real part.
4. Identify the stability type of the closed-loop origin (e.g. stable node, stable focus).
5. State whether the stabilization is **local** or **global**, and justify.

---

## Task 3 — Lyapunov function via Sum-of-Squares (SoS)

1. Propose a Lyapunov function candidate $V(x)$, e.g.
   $$
   V(x) = x^\top P x, \qquad P = P^\top \succ 0
   $$
   or a polynomial $V(x)$ of higher degree if needed.
2. Formulate the search for $V$ as a **Sum-of-Squares (SoS)** feasibility problem:
   $$
   \begin{aligned}
   & \text{find } V(x) \text{ such that} \\
   & V(x) - \epsilon \|x\|^2 \in \Sigma[x], \\
   & -\dot{V}(x) - \epsilon \|x\|^2 \in \Sigma[x] \quad \text{(locally or globally)}
   \end{aligned}
   $$
   where $\Sigma[x]$ denotes the set of sum-of-squares polynomials.
3. For **local** stability, add a sublevel-set constraint, e.g.
   $$
   V(x) \leq \rho \implies -\dot{V}(x) > 0
   $$
   and formulate it with an SoS multiplier (Positivstellensatz / S-procedure).
4. Solve the SoS program and report the resulting $V(x)$.
5. Conclude whether the Lyapunov function proves **local** or **global** asymptotic stability of the closed-loop origin.

# Problem 2. SoS Stabilization of an Inverted Pendulum

Consider an inverted pendulum, where $\theta$ is the angle between the vertical and the pole.

---

## System Description

Let:

- $\theta$ — pole angle from the vertical,
- $\dot{\theta}$ — pole angular velocity,
- $u$ — control torque applied at the pivot,
- $m$ — pole mass,
- $l$ — distance from the pivot to the pole's center of mass,
- $g$ — gravitational acceleration,
- $b$ — viscous damping coefficient at the pivot.

The nonlinear equation of motion is:

$$
m l^2 \, \ddot{\theta} + b \, \dot{\theta} - m g l \sin\theta = u
$$

Define the state vector

$$
z = \begin{pmatrix} x_1 \\ x_2 \end{pmatrix} = \begin{pmatrix} \theta \\ \dot{\theta} \end{pmatrix}.
$$

---

## Task 1 — Linearization and linear control law

1. Rewrite the pendulum dynamics in state-space form $\dot{z} = f(z, u)$ with $u = 0$.
2. Linearize around the origin $z = 0$ to obtain
   $$
   \dot{z} = A z + B u.
   $$
3. Find a linear control law
   $$
   u = -K z = -k_1 x_1 - k_2 x_2
   $$
   such that the closed-loop matrix $A_c = A - BK$ is Hurwitz.
4. Report the gains $k_1, k_2$ and the eigenvalues of $A_c$.
5. Identify the stability type of the closed-loop origin.

---

## Task 2 — Nonlinear closed-loop model and trig handling

Substitute $u = -K z$ into the nonlinear pendulum equation. To handle the trigonometric nonlinearity, choose **one** of the following and state why:

1. **Small-angle approximation:** $\sin\theta \approx \theta$.
2. **Truncated Taylor expansion:** e.g. $\sin\theta \approx \theta - \frac{\theta^3}{6}$.
3. **Change of variables:** introduce $s = \sin\theta$, $c = \cos\theta$, with the algebraic constraint $s^2 + c^2 = 1$, and treat $(s, c, \dot{\theta})$ as the state.

Write the resulting **polynomial** closed-loop system
$$
\dot{z} = f_c(z)
$$
that will be used in the SoS formulation.

---

## Task 3 — Lyapunov function search via SoS

1. Propose a Lyapunov function candidate $V(z)$, e.g. a quadratic form
   $$
   V(z) = z^\top P z, \qquad P = P^\top \succ 0,
   $$
   or a higher-degree polynomial if needed.
2. Formulate the search for $V$ as an **SoS feasibility problem**:
   $$
   \begin{aligned}
   & \text{find } V(z) \text{ such that} \\
   & V(z) - \epsilon \|z\|^2 \in \Sigma[z], \\
   & -\dot{V}(z) - \epsilon \|z\|^2 \in \Sigma[z] \quad \text{(locally, on a sublevel set } V(z) \leq \rho \text{)}.
   \end{aligned}
   $$
3. If you used the $s = \sin\theta,\ c = \cos\theta$ change of variables, include the constraint
   $$
   s^2 + c^2 = 1
   $$
   in the SoS program (e.g. as an equality constraint or via a multiplier).
---

## Task 4 — Solve and show the certificate

1. Implement the SoS program, e.g. using `SumOfSquares.jl`.
2. Solve the feasibility problem and **show that the solver returns a feasible solution** (a certificate)
3. Comment on whether the certificate proves **local** or **global** stability.
