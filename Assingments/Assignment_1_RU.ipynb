# Задача 1. Локальная устойчивость и стабилизация полиномиальных систем линейной обратной связью

## Задание 1 — Локальная устойчивость начала координат (разомкнутая система, $u = 0$)

Рассмотрим систему при $u = 0$:

$$
\left\{
\begin{aligned}
\dot{x}_1 &= f_1(x_1, x_2) \\
\dot{x}_2 &= f_2(x_1, x_2)
\end{aligned}
\right.
$$

1. Вычислите матрицу Якоби

   $$
   A = \left.\frac{\partial f}{\partial x}\right|_{x=0}
   $$

2. Вычислите собственные значения матрицы $A$.

3. Используя критерий по собственным значениям, определите, является ли начало координат $x_1 = 0,\ x_2 = 0$:

   * асимптотически устойчивым,
   * нейтрально устойчивым,
   * или неустойчивым.

## Задание 2 — Стабилизация линейной обратной связью

Введём линейный закон управления вида

$$

1. Подставьте $u$ в систему и получите замкнутую динамику.

2. Вычислите матрицу Якоби замкнутой системы $A_c$ в начале координат.

3. Выберите коэффициенты $k_1, k_2$ (если это возможно) так, чтобы $A_c$ была гурвицевой, то есть все собственные значения имели строго отрицательные вещественные части.

4. Определите тип устойчивости замкнутой системы в начале координат (например, устойчивый узел, устойчивый фокус).

5. Укажите, является ли стабилизация **локальной** или **глобальной**, и обоснуйте свой ответ.

## Задание 3 — Функция Ляпунова методом сумм квадратов (SoS)

1. Предложите кандидат на функцию Ляпунова $V(x)$, например

   $$
   V(x) = x^\top P x, \qquad P = P^\top \succ 0
   $$

   или полиномиальную функцию $V(x)$ более высокой степени, если это необходимо.

2. Сформулируйте поиск $V$ как задачу на существование **Sum-of-Squares (SoS)**:

   $$
   \begin{aligned}
   & \text{найти } V(x) \text{ такую, что} \\
   & V(x) - \epsilon \|x\|^2 \in \Sigma[x], \\
   & -\dot{V}(x) - \epsilon \|x\|^2 \in \Sigma[x] \quad \text{(локально или глобально)}
   \end{aligned}
   $$

   где $\Sigma[x]$ обозначает множество полиномов, представимых в виде сумм квадратов.

3. Для **локальной** устойчивости добавьте ограничение на подуровневое множество, например

   $$
   V(x) \leq \rho \implies -\dot{V}(x) > 0
   $$

   и сформулируйте его с использованием SoS-множителя (Positivstellensatz / S-procedure).

4. Решите SoS-задачу и приведите полученную функцию $V(x)$.

5. Сделайте вывод о том, доказывает ли найденная функция Ляпунова **локальную** или **глобальную** асимптотическую устойчивость начала координат замкнутой системы.

# Задача 2. Стабилизация перевёрнутого маятника методом SoS

Рассмотрим перевёрнутый маятник, где $\theta$ — угол между вертикалью и стержнем.

---

## Описание системы

Пусть:

* $\theta$ — угол отклонения стержня от вертикали,

* $\dot{\theta}$ — угловая скорость стержня,

* $u$ — управляющий момент, приложенный к точке подвеса,

* $m$ — масса стержня,

* $l$ — расстояние от точки подвеса до центра масс стержня,

* $g$ — ускорение свободного падения,

* $b$ — коэффициент вязкого трения в точке подвеса.

Нелинейное уравнение движения имеет вид:

$$
m l^2 \, \ddot{\theta} + b \, \dot{\theta} - m g l \sin\theta = u
$$

Определим вектор состояния

$$
z = \begin{pmatrix} x_1 \\ x_2 \end{pmatrix}
= \begin{pmatrix} \theta \\ \dot{\theta} \end{pmatrix}.
$$

---

## Задание 1 — Линеаризация и линейный закон управления

1. Перепишите динамику маятника в пространстве состояний $\dot{z} = f(z, u)$ при $u = 0$.

2. Линеаризуйте систему в окрестности начала координат $z = 0$ и получите

   $$
   \dot{z} = A z + B u.
   $$

3. Найдите линейный закон управления

   $$
   u = -K z = -k_1 x_1 - k_2 x_2
   $$

   такой, чтобы замкнутая матрица $A_c = A - BK$ была гурвицевой.

4. Приведите найденные коэффициенты $k_1, k_2$ и собственные значения матрицы $A_c$.

5. Определите тип устойчивости начала координат замкнутой системы.

---

## Задание 2 — Нелинейная замкнутая модель и работа с тригонометрической нелинейностью

Подставьте $u = -K z$ в нелинейное уравнение движения маятника. Для работы с тригонометрической нелинейностью выберите **один** из следующих подходов и объясните свой выбор:

1. **Приближение для малых углов:** $\sin\theta \approx \theta$.

2. **Обрезанное разложение Тейлора:** например, $\sin\theta \approx \theta - \frac{\theta^3}{6}$.

3. **Замена переменных:** введите $s = \sin\theta$, $c = \cos\theta$ с алгебраическим ограничением $s^2 + c^2 = 1$ и рассматривайте $(s, c, \dot{\theta})$ как вектор состояния.

Запишите полученную **полиномиальную** замкнутую систему

$$
\dot{z} = f_c(z)
$$

которая будет использоваться в SoS-формулировке.

---

## Задание 3 — Поиск функции Ляпунова методом SoS

1. Предложите кандидат на функцию Ляпунова $V(z)$, например квадратичную форму

   $$

   или полином более высокой степени, если это необходимо.

2. Сформулируйте поиск $V$ как **задачу на существование SoS-сертификата**:
